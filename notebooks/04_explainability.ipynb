# 04 - Explainability: SHAP & LIME

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

import joblib
import pandas as pd

from src.explain import explain_with_lime, explain_with_shap, top_shap_features_for_instance
from src.preprocessing import preprocess_dataframe

model = joblib.load('../models/model.pkl')
df = pd.read_csv('../data/raw/fake_job_postings.csv')
sample = df.sample(min(200, len(df)), random_state=42)
clean_sample = preprocess_dataframe(sample)
```

## Global explanation with SHAP

```python
shap_values, feature_names = explain_with_shap(model, clean_sample['text'].tolist())
import shap
shap.summary_plot(shap_values, feature_names=feature_names, show=True)
```

## Local explanation for a single fake-labeled example (LIME)

```python
fake_example = clean_sample[clean_sample['fraudulent'] == 1].iloc[0]['text']
lime_list, lime_exp = explain_with_lime(model, fake_example, num_features=12)
for word, weight in lime_list:
    print(f'{word:20s} {weight:+.4f}')
```

```python
# Render inline in the notebook
lime_exp.show_in_notebook(text=True)
```
