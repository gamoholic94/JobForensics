# 01 - Exploratory Data Analysis
Fake Job Posting Detection using Explainable AI

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.preprocessing import load_raw_dataset

DATA_PATH = os.path.join('..', 'data', 'raw', 'fake_job_postings.csv')
df = load_raw_dataset(DATA_PATH)
df.head()
```

```python
df.info()
df.isna().sum().sort_values(ascending=False)
```

```python
# Class balance
print(df['fraudulent'].value_counts(normalize=True))
sns.countplot(x='fraudulent', data=df)
plt.title('Class distribution (0 = Real, 1 = Fake)')
plt.savefig('../reports/figures/class_distribution.png', bbox_inches='tight')
plt.show()
```

```python
# Text length analysis
df['description_len'] = df['description'].fillna('').apply(len)
sns.boxplot(x='fraudulent', y='description_len', data=df)
plt.title('Description length by class')
plt.savefig('../reports/figures/description_length_by_class.png', bbox_inches='tight')
plt.show()
```

```python
# Company logo / telecommute correlation with fraud
for col in ['has_company_logo', 'telecommuting', 'has_questions']:
    if col in df.columns:
        print(df.groupby('fraudulent')[col].mean())
        print()
```
